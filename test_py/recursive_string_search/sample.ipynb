```python
import os
```

```python
def recursive_search(dirpath):
    curdir = os.listdir(dirpath)
    for file in curdir:
        if os.path.isfile(file):
            print("file: "+file)
            if file.endswith('.ini'):
                print(file)
        elif os.path.isdir(file):
            print("dir: " + file)
            recursive_search(os.path.abspath(file))
```

```python
recursive_search(os.path.curdir)
```
